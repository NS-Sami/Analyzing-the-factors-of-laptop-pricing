```python
import pandas as pd

df = pd.read_csv('laptop_price - dataset.csv')  # Adjust filename if needed
df.head()
```

```python
df.info()
df.describe()
df.isnull().sum()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,6))
sns.countplot(data=df, x='Company')
plt.title('Laptop Count by Company')
plt.xticks(rotation=45)
plt.show()
```

```python
df['TypeName'].value_counts().plot.pie(autopct='%1.1f%%', figsize=(6,6))
plt.title('Distribution of Laptop Types')
plt.ylabel('')
plt.show()
```

```python
df_sorted = df.sort_values(by='CPU_Frequency (GHz)')
plt.plot(df_sorted['CPU_Frequency (GHz)'], df_sorted['Price (Euro)'])
plt.xlabel('CPU Frequency (GHz)')
plt.ylabel('Price (Euro)')
plt.title('Price vs CPU Frequency')
plt.grid(True)
plt.show()

```

```python
sns.boxplot(data=df, x='Company', y='Price (Euro)')
plt.xticks(rotation=45)
plt.title('Laptop Prices by Company')
plt.show()

```

```python
# Correlation heatmap
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Correlation Between Numeric Features')
plt.show()
```
